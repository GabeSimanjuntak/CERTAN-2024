```python
# Mengimpor NumPy untuk pengolahan data numerik.
import numpy as np

# Mengimpor Pandas untuk mengolah data dalam bentuk tabel.
import pandas as pd

# Mengimpor fungsi load_iris untuk mengambil dataset Iris dari scikit-learn.
from sklearn.datasets import load_iris

# Memuat dataset Iris dari scikit-learn.
iris = load_iris()
```

```python
# Mengubah data Iris menjadi DataFrame Pandas dengan nama kolom sesuai fitur.
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Menambahkan kolom Species yang berisi nama spesies Iris.
df["Species"] = iris.target_names[iris.target]

# Menentukan X sebagai data fitur dengan menghapus kolom Species.
X = df.drop(columns=["Species"])

# Menentukan y sebagai target atau label Species.
y = df["Species"]

# Menampilkan data fitur X.
print(X)
```

```python
# Menampilkan data target atau label Species.
y
```

```python
# Mengimpor LabelEncoder untuk mengubah label kategori menjadi angka.
from sklearn.preprocessing import LabelEncoder

# Membuat objek LabelEncoder.
lab = LabelEncoder()

# Mengubah label Species menjadi nilai numerik.
y = lab.fit_transform(y)

# Menampilkan hasil encoding label.
y
```

```python
# Mengimpor fungsi train_test_split untuk membagi data menjadi data training dan testing.
from sklearn.model_selection import train_test_split

# Membagi data fitur X menjadi 80% data training dan 20% data testing.
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)

# Membagi data target y menjadi 80% data training dan 20% data testing.
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

```python
# Mengimpor GaussianNB untuk membuat model Naive Bayes.
from sklearn.naive_bayes import GaussianNB

# Membuat objek model Gaussian Naive Bayes.
model = GaussianNB()

# Melatih model menggunakan data training X_train dan y_train.
NB_model = model.fit(X_train, y_train)
```

```python
# Menggunakan model Naive Bayes untuk memprediksi data testing.
result = NB_model.predict(X_test)

# Menampilkan hasil prediksi.
result
```

```python
# Mengimpor accuracy_score untuk menghitung akurasi hasil prediksi.
from sklearn.metrics import accuracy_score

# Menghitung dan menampilkan akurasi data testing dalam bentuk persen.
print('Accuracy of data testing: ', accuracy_score(y_test, result) * 100)
```

```python
# Menentukan data baru yang akan diprediksi.
input_data = [[5.1, 3.5, 1.4, 0.2]]

# Melakukan prediksi menggunakan model Naive Bayes.
result = NB_model.predict(input_data)

# Mengubah hasil prediksi angka menjadi nama species.
prediction_label = lab.inverse_transform(result)

# Menampilkan hasil prediksi untuk data baru.
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
