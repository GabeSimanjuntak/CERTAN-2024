```python
#operasi elementwise (mengoperasikan setiap elemen yang berkorespondesnsi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a+b
hasil2 = a-b
hasil3 = a*b
hasil4 = a/b
hasil5 = a**b

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
#operator Elementwise ada masalah di perkalian matriks karena mengalikan per elemen
a = np.array( [(1, 2, 3), (4, 5, 6)] )
b = np.array( [(3, 4, 5), (3, 4, 5)] )

hasil = a*b
print("hasil =", hasil)
```

```python
#mengetahui ukuran matriks
a = np.array( [(1,2,3),(4,5,6)] )
print("matriks a berukuran =", a.shape)

#transpose matriks
print("transpose matriks a =", a.transpose())
print("transpose matriks a =", np.transpose(a))
print("transpose matriks a =", a.T)

#Pada Machine learning biasanya flatten dan stacking sangat penting
#flatten array, vektor baris
print("flatten dari matriks a =", a.ravel())
```

```python
#invers matriks
a = np.array( [(1,-1), (1,1)] )
print(a)

a_inv = np.linalg.inv(a)
print(a_inv)

det_a = np.linalg.det(a)
print(det_a)
```

```python
#solusi persamaan linear
A = np.array( [(1,-6),(2,-5)] )
Y = np.array([-35, 21])
print(A)
print(Y)

A_inv = np.linalg.inv(A)

#menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

#cara lain
X2 = np.linalg.solve(A,Y)
print(X2)
```

```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X,Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
#Soal 2
x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, y)

# Persamaan: x^6 + 3*x^4*y^2 - 3*x^4 + 3*x^2*y^4 - x^2*y^3 - 6*x^2*y^2 + 3*x^2 + y^6 - 3*y^4 + 3*y^2 = 1
F = (X**6 + 3*(X**4)*(Y**2) - 3*(X**4) + 3*(X**2)*(Y**4) 
     - (X**2)*(Y**3) - 6*(X**2)*(Y**2) + 3*(X**2) 
     + Y**6 - 3*(Y**4) + 3*(Y**2))

plt.figure(figsize=(6, 6))
plt.contour(X, Y, F, levels=[1], colors='red')
plt.title("Grafik 2D Persamaan (Bentuk Hati)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()
```

```python
#Soal 3
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)

Z = ((5 * Y / 4) - np.abs(X))**2 + X**2

fig = plt.figure(figsize=(8, 6))
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

ax.set_title("Grafik 3D Persamaan")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
